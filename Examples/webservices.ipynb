# WFS Example

```python
import requests
import geopandas as gpd

from IPython.display import Image
```

```python
url = """
https://ahocevar.com/geoserver/wfs?
service=WFS
&version=1.1.0
&request=GetFeature
&typename=osm:water_areas
&outputFormat=application/json
&srsname=EPSG:3857
&bbox=-8933308.907933006,5370452.51819444,-8884465.646858776,5393383.626679991,EPSG:3857
"""

req = requests.get(url)

req.status_code
```

```python
req.json()
```

```python
gdf = gpd.GeoDataFrame.from_features(req.json(), crs="EPSG:3857")
gdf.plot()
```
